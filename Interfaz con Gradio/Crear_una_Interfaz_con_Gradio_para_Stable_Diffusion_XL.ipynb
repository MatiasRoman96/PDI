```python
from diffusers import StableDiffusionXLPipeline
import torch
from PIL import Image
import gradio as gr
```

```python
pipe = StableDiffusionXLPipeline.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16
)
pipe.to("cuda")
```

```python
def generar_imagen(prompt, negative_prompt, steps, guidance):
    images = pipe(
        prompt=prompt,
        negative_prompt=negative_prompt,
        num_inference_steps=steps,
        guidance_scale=guidance
    ).images
    return images[0]

```

```python
gr.Interface(
    fn=generar_imagen,
    inputs=gr.Textbox(label="Prompt para la imagen"),
    outputs=gr.Image(type="pil", label="Imagen generada"),
    title="Generador de Imágenes con Stable Diffusion XL",
    description="Escribí una descripción y generá una imagen con IA usando Stable Diffusion XL."
).launch()
```

```python
gr.Interface(
    fn=generar_imagen,
    inputs=[
        gr.Textbox(label="Prompt para la imagen"),
        gr.Slider(10, 100, value=50, label="Número de pasos de inferencia"),
        gr.Slider(1.0, 15.0, value=7.5, step=0.1, label="Guidance scale")
    ],
    outputs=gr.Image(type="pil", label="Imagen generada"),
    title="Generador de Imágenes con Stable Diffusion XL",
    description="Escribí una descripción, ajustá los parámetros y generá una imagen con IA usando Stable Diffusion XL."
).launch()
```
